```python
import pandas as pd
import os

# Coloque aqui o caminho para a pasta onde você descompactou seus logs
log_directory = './'


for filename in os.listdir(os.getcwd()):
    if ".log" in filename and ".csv" not in filename:
        log_file = filename # O nome do seu arquivo de log
        print(log_file)
        full_path = os.path.join(log_directory, log_file)
        
        # Este bloco de código abre o arquivo, encontra a linha #fields e extrai os nomes das colunas
        columns = []
        with open(full_path, 'r') as f:
            for line in f:
                if line.startswith('#fields'):
                    # Separa a linha por TAB e pega todos os nomes a partir do segundo elemento
                    columns = line.strip().split('\t')[1:]
                    break
        
        # --- LEITURA DO ARQUIVO COM PANDAS ---
        try:
            df_conn = pd.read_csv(
                full_path,
                sep='\t',          # Informamos que o separador é TAB
                names=columns,       # Fornecemos os nomes das colunas que extraímos
                comment='#',       # Ignora todas as linhas de metadados que começam com '#'
                header=None,       # Diz ao pandas que não há uma linha de cabeçalho nos dados que ele está lendo
                low_memory=False
            )
        
            print("Log lido com sucesso! O DataFrame está pronto.")
            print(f"Número de conexões capturadas: {len(df_conn)}")
            
            # Exibe as 5 primeiras linhas do DataFrame organizado
            print("\nAmostra do DataFrame:")
            df_conn['attack_cat'] = "Reconnaissance"  # TROCAR O ATAQUE APÓS COLETA
            df_conn['label'] = 1
            display(df_conn.head())
            df_conn.to_csv(filename + '.csv', index=False)
        
        except FileNotFoundError:
            print(f"Erro: Arquivo não encontrado em '{full_path}'. Verifique o caminho.")
        except Exception as e:
            print(f"Ocorreu um erro: {e}")
```
